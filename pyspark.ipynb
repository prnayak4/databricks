### **Difference Between dropDuplicates() vs distinct()
### distinct) → removes duplicate rows (entire row match).
### dropDuplicates([subset]) → removes duplicates based on one or more specific columns.**

```python
data = [
    (1, "Alice", 23),  # duplicate row
    (2, "Bob", 34),  # duplicate row
    (3, "Charlie", 29),
    (1, "Alice", 23),  # duplicate row
    (2, "Bob", 34),  # duplicate row
    (6, "Alice", 30)  # same name, different age
]
columns = ["id", "name", "age"]
df = spark. createDataFrame(data, columns)
df. show()
```

```python
display(df.distinct())

```

```python
display(df.dropDuplicates())
```

```python
display(df.dropDuplicates(['age']))
```

**### What is groupBy() in PySpark?
### The groupBy transformation groups rows that have the same values in specified column(s).
### After grouping, you must apply aggregation functions (like count, sum, avg, min, max).
### Similar to GROUP BY in SQL.**

```python
data = [
(1, "Alice", "NY", 2000),
(2, "Bob", "CA", 1500), 
(3, "Charlie", "NY", 3000),
 (4, "David", "CA", 2500),
(5, "Eve", "TX", 1800), 
(6, "Frank", "TX", 2200),
]
columns = ["id", "name", "state", "salary"]
df = spark. createDataFrame(data, columns)
display(df)
```

```python
display(df.groupBy('state').count())
display(df.groupBy('state').sum('salary'))
display(df.groupBy('state').avg('salary'))
```

```python
display(df.groupBy('state').agg({'salary' :'sum'}))
```

```python
display(df.groupBy('state').agg({'salary' :'min'}))
```

```python
display(df.groupBy('state').agg({'salary' :'max'}))
```

```python
display(df.groupBy('state').agg({'salary' :'avg'}))
```

### **pyspark tutorial
### join in pyspark
### PySpark Join is used to combine two DataFrames
### PySpark's join operation combines data from two or more Datasets based on a common column or key. It is a fundamental operation in PySpark and is similar to SQL joins.
### it supports all basic join type operations available in traditional SQL like INNER, LEFT OUTER, RIGHT OUTER, LEFT ANTI, LEFT SEMI, CROSS, SELF JOIN
### inner Join: Returns only the rows with matching keys in both DataFrames.
### Left Join: Returns all rows from the left DataFrame and matching rows from the right DataFrame.
### Right Join: Returns all rows from the right DataFrame and matching rows from the left DataFrame.
### Full Outer Join: Returns all rows from both DataFrames, including matching and non-matching rows.
### Left Semi Join: Returns all rows from the left DataFrame where there is a match in the right DataFrame.
### Left Anti Join: Returns all rows from the left DataFrame where there is no match in the right DataFrame.**

**What is a Join in PySpark?**

**### PySpark UDF ( User Defined Function)
### is the most useful feature of Spark SQL & DataFrame that is used to extend the PySpark build in capabilities.
### UD's are used to extend the functions of the framework and re-use these functions on multiple
### DataFrame's
### For example, you wanted to convert every first letter of a word in a name string to a capital case;
### PySpark build-in features don't have this function hence you can create it a UDF and reuse this as needed on many Data Frames
### Note: UDF's are the most expensive operations hence use them only you have no choice and when essential**

```python
def age_grp (age) :
    if age is None:
        return "UNKNOWN"
    elif age < 18:
        return "MINOR"
    elif age >= 18 and age < 40:
        return "ADULT"
    else:
        return "SENIOR"

age_grp(40)
```

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType
age_grp_category= udf (age_grp, StringType ( ))
```

```python
df.withColumn('age_cat',age_grp_category(df.age)).show()
```

**ROW_NUMBERO
& Assigns a unique sequential number to each row within a partition (group), ordered by a column.
Even if values are the same, each row gets a different number. M Example: 1,2,3,4...**

```python
from pyspark. sql.window import Window 
from pyspark. sql. functions import *
w = Window.partitionBy('dept'). orderBy(col('salary'). desc())
df.withColumn('rn', row_number().over(w)).show()
```

### RANK(
### Assigns a rank to rows within a partition based on order.
### Ties get the same rank.
### But the next rank will skip numbers.
### Example: 1, 2, 2, 4 (notice skip of 3).

```python
from pyspark. sql.window import Window 
from pyspark. sql. functions import *
w = Window.partitionBy('dept'). orderBy(col('salary'). desc())
df.withColumn('rn', rank().over(w)).show()
```

## **DENSE_RANK(
## • Similar to RANK, but it does not skip numbers after ties.
## Ties get the same rank.
## Next rank is consecutive. M Example: 1, 2, 2, 3 (no skip).**
 


```python
from pyspark. sql.window import Window 
from pyspark. sql. functions import *
w = Window.partitionBy('dept'). orderBy(col('salary'). desc())
df.withColumn('rn', dense_rank().over(w)).show()
```

## **LAG(column, n)
## Fetches the value from the previous row (n steps back) in the same partition.
## Useful for comparing with "previous record". Example: Salary difference from last employee.**

```python
from pyspark. sql.window import Window 
from pyspark. sql. functions import *
w = Window.partitionBy('dept'). orderBy(col('salary'). desc())
df.withColumn('rn', lag('salary', offset=1).over(w)).show()
```

### LEAD(column, n)
### Fetches the value from the next row (n steps ahead) in the same partition.
### Useful for comparing with "next record". Example: Salary gap with next higher salary.

```python
from pyspark. sql.window import Window 
from pyspark. sql. functions import *
w = Window.partitionBy('dept'). orderBy(col('salary'). desc())
df.withColumn('rn', lead('salary', offset=1).over(w)).show()
```
