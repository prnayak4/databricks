```python
df = spark. read.csv("/Volumes/development/lake/lake_practical/sales/", header=True)
display (df)
```

```python
from pyspark. sql. functions import to_date, col, expr, try_to_date
df_clean = (df. withColumn("order_date", try_to_date(col("order_date"), "yyyy-MM-dd"))
.withColumn("quantity", expr("try_cast(price as decimal(12,2))"))
.dropna(subset=["order_id", "quantity"]))
# Write to Silver Delta Table
(df_clean.write.format("delta")
.mode("overwrite")
.saveAsTable("development.lake.sales_orders"))
```
